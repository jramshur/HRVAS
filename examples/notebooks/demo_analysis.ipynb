# HRVAS 2.0 Demo

This notebook demonstrates the functionality of the Python port of HRVAS.

```python
import sys
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure we can import hrvas
sys.path.append(os.path.abspath('../../'))

from hrvas.io import load_ibi

# Set style
sns.set_theme(style="whitegrid")
```

```python
# Load synthetic data
file_path = '../../tests/fixtures/sample_data/sine_0.25hz.csv'
data = load_ibi(file_path)

print(f"Loaded {len(data)} beats.")
data.head()
```

```python
# Plot IBI series
plt.figure(figsize=(10, 4))
plt.plot(data['Time'], data['IBI'], label='IBI')
plt.xlabel('Time (s)')
plt.ylabel('IBI (s)')
plt.title('Synthetic IBI Series (0.25 Hz RSA)')
plt.legend()
plt.show()
```
